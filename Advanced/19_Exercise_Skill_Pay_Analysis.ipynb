```python
# Importing Libraries
import ast
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt  

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
df['job_skills'] = df['job_skills'].apply(lambda x: ast.literal_eval(x) if pd.notna(x) else(x))
```

```python
df_DA_US = df[(df['job_country'] == 'United States')&(df['job_title_short']=='Data Analyst')].copy()

df_DA_US = df_DA_US.dropna(subset=['salary_year_avg'])
```

```python
df_DA_US=df_DA_US.explode('job_skills')
```

```python
df_DA_US[['salary_year_avg', 'job_skills']]
```

```python
df_DA_US_group = df_DA_US.groupby('job_skills')['salary_year_avg'].agg(['count', 'median'])
```

```python
df_DA_top_pay = df_DA_US_group.sort_values(by='median', ascending=False).head(10)
```

```python
df_DA_top_pay
```

```python
df_DA_skills = df_DA_US_group.sort_values(by='count', ascending=False).head(10).sort_values(by='median', ascending=False)
df_DA_skills
```

```python
fig, ax = plt.subplots(2,1)

df_DA_top_pay.plot(kind='barh', y='median', ax=ax[0], legend=False)
ax[0].invert_yaxis()
ax[0].set_xlabel('')
ax[0].set_ylabel('')
ax[0].set_title('Top 10 Highest Paid Skills for Data Analystics')
ax[0].xaxis.set_major_formatter(plt.FuncFormatter(lambda x,_: f'${int(x/1000)}K'))


df_DA_skills[::-1].plot(kind='barh', y='median', ax=ax[1], legend=False, xlabel='Median Salary ($USD)', ylabel='', title='Top 10 Most in Demand Skills for Data Analystics')
ax[1].set_xlim(ax[0].get_xlim())
ax[1].xaxis.set_major_formatter(plt.FuncFormatter(lambda x,_: f'${int(x/1000)}K'))
fig.tight_layout()
```
