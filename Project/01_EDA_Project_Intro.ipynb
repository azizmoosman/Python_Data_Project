```python
# Importing Libraries
import ast
import pandas as pd
import seaborn as sns
from datasets import load_dataset
import matplotlib.pyplot as plt  

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
df['job_skills'] = df['job_skills'].apply(lambda x: ast.literal_eval(x) if pd.notna(x) else(x))
```

```python
df_DA_SU = df[(df['job_country']== 'United States')&(df['job_title_short']=='Data Analyst')]
```

```python
df_DA_SU
```

```python
df_plot=df_DA_SU['job_location'].value_counts().head(10).to_frame()


sns.set_theme(style='ticks')
sns.barplot(data=df_plot, x='count', y='job_location', hue='count', palette='dark:b_r', legend=False)
sns.despine()
plt.title('Number of Data Analyst Jobs in Sudan')
plt.xlabel('Number of Jobs')
plt.ylabel('')
plt.show()
```

```python
fig,ax=plt.subplots(1,3)

dict_coloumn= {
    'job_work_from_home': 'Work from Home',
    'job_no_degree_mention': 'Job Degree Req',
    'job_health_insurance': 'Health Insurance Offered'
}

for i, (column, title) in enumerate(dict_coloumn.items()):
    ax[i].pie(df_DA_SU[column].value_counts(),startangle=90, autopct='%1.1f%%', labels=['False','True'])
    ax[i].set_title(title)

plt.show()
```

```python
df_plot=df_DA_SU['company_name'].value_counts().head(10).to_frame()


sns.set_theme(style='ticks')
sns.barplot(data=df_plot, x='count', y='company_name', hue='count', palette='dark:b_r', legend=False)
sns.despine()
plt.title('Counts of Companies posting Analyst Jobs in Sudan')
plt.xlabel('Number of Jobs')
plt.ylabel('')
plt.show()
```
